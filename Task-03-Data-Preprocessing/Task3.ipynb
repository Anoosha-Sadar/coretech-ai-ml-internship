```python
import pandas as pd

data = {
    'CustomerID':[1,2,3,4,4,5],
    'Age':[25,30,22,35,35,28],
    'Gender':['Male','Female','Male','Female','Female','Male'],
    'City':['Karachi','Hyderabad','Karachi','Sukkur','Sukkur','Nawabshah'],
    'Income':[50000,60000,None,70000,70000,55000],
    'Purchased':['Yes','No','Yes','No','No','Yes']
}

df = pd.DataFrame(data)

df.to_csv('dataset.csv',index=False)

print("dataset.csv created successfully")
```

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.feature_selection import SelectKBest, f_classif
import matplotlib.pyplot as plt

df = pd.read_csv("dataset.csv")

df["Income"].fillna(df["Income"].mean(), inplace=True)

df.drop_duplicates(inplace=True)

le = LabelEncoder()
df["Gender"] = le.fit_transform(df["Gender"])
df["Purchased"] = le.fit_transform(df["Purchased"])

df = pd.get_dummies(df, columns=["City"])

scaler = MinMaxScaler()
df[["Age","Income"]] = scaler.fit_transform(df[["Age","Income"]])

X = df.drop("Purchased", axis=1)
y = df["Purchased"]

selector = SelectKBest(score_func=f_classif, k=3)
selector.fit(X,y)

df.to_csv("cleaned_dataset.csv", index=False)

before_rows = 6
after_rows = len(df)

plt.figure(figsize=(6,4))
plt.bar(["Before","After"], [before_rows, after_rows])
plt.title("Dataset Comparison")
plt.savefig("before_after_comparison.png")
plt.show()

print("Task Completed")
```
